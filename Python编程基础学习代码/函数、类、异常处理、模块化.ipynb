```python
#------作用域（全局变量和局部变量）------
x = 10

def foo():
    #global x
    x = x + 40     # ← 这行会报错！因为函数内部属于局部，局部变量需要重新赋值，除非明确调用 global x
    print(x)

foo()
# ❌ UnboundLocalError: local variable 'x' referenced before assignment
```

```python
#------类和对象------
class Phonebook:                    # ① class 定义类（设计图）
    def __init__(self):             # ② __init__：造对象时自动跑的"出厂初始化"
        self.contacts = {}          # ③ self = 对象自己；这行是"对象自己的通讯录"

    def add(self, name, phone):     # ④ 方法=写在类里的函数，第一个参数永远是 self
        self.contacts[name] = phone

    def search(self, name):
        if name in self.contacts:
            return self.contacts[name]
        return None

    def show_all(self):
        for name, phone in self.contacts.items():
            print(f"{name}: {phone}")
work = Phonebook()
private = Phonebook()

work.add("张三", "13800000000")
private.add("李四", "13900000000")

private.show_all()
work.show_all()
```

```python
#------完整版本（类+try-except）------
class Phonebook:
    def __init__(self,name="默认"):
        self.book_name = name
        self.contacts = {}

    def add(self,name,phone):
        self.contacts[name] = phone
        return f"[{self.book_name}]已添加：{name}"

    def search(self,name):
        try:
            return self.contacts[name]
        except KeyError:
            return f"[{self.book_name}]没有{name}"

    def delete(self,name):
        try:
            del self.contacts[name]
            return f"[{self.book_name}]已删除：{name}"
        except KeyError:
            return f"[{self.book_name}]没有{name},无法删除"

    def show_all(self):
        print(f"==={self.book_name}===")
        if not self.contacts:
            print("空")
        for name,phone in self.contacts.items():
            print(f"{name}:{phone}")

            
# ===== 造两个对象，各干各的 =====
work = Phonebook("工作通讯录")
private = Phonebook("私人通讯录")

work.add("张三", "13800000000")
private.add("闺蜜", "13900000000")

work.show_all()
private.show_all()

print(work.search("张三"))
print(work.search("王五"))      # ← 触发 try-except
print(work.delete("李四"))      # ← 触发 try-except
```
