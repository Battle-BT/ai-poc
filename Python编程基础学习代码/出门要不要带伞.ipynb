```python
def need_umbrella(weather, temperature):
    """判断：下雨带伞；阴天且低于25度也带伞；其他不用"""
    if weather == "下雨" or weather == "下雪":
        return True
    elif weather == "阴天" and temperature < 25:
        return True
    else:
        return False

cases = [("下雨", 22), ("晴天", 30), ("阴天", 20),("下雪",40)]

for weather, temperature in cases:
    if need_umbrella(weather, temperature):
        print(f"{weather} {temperature}度 → 带伞！")
    else:
        print(f"{weather} {temperature}度 → 不用带")
```
