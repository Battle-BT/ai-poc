```python
#------requests(GET)------
import requests

url = "https://api.open-meteo.com/v1/forecast?latitude=22.5&longitude=114&current=temperature_2m,wind_speed_10m"

try:
    resp = requests.get(url, timeout=10)
    resp.raise_for_status()  # 不是200就报错
    
    print("======网络状态======")
    print(resp.status_code)
    
    print("======全部文本信息=======")
    print(resp.text)
    
    data = resp.json()
    print("======以下是json格式======")
    print(f"JSON{data}")
    
    # 层层索取
    temp = data["current"]["temperature_2m"]
    wind = data["current"]["wind_speed_10m"]
    
    print(f"深圳现在温度是{temp}°")
    print(f"风速为{wind}km/h")

except requests.exceptions.ConnectionError:
    print("网络连不上，请检查网络")
except requests.exceptions.Timeout:
    print("请求超时，请稍后重试")
except Exception as e:
    print(f"出错了：{e}")
```

```python
#-------函数封装------
import requests

def get_weather(lat, lon):
    """输入经纬度，输出天气"""
    try:
        resp = requests.get(
            "https://api.open-meteo.com/v1/forecast",
            params={"latitude": lat, "longitude": lon, "current": "temperature_2m,wind_speed_10m"},
            timeout=10
        )
        resp.raise_for_status()
        data = resp.json()
        
        temp = data["current"]["temperature_2m"]
        wind = data["current"]["wind_speed_10m"]
        
        print(f"温度：{temp}°C")
        print(f"风速：{wind} km/h")
        
    except Exception as e:
        print(f"查天气失败：{e}")

# 查深圳
get_weather(22.5, 114.0)

# 查北京
get_weather(39.9, 116.4)
```

```python
#------requests(POST)------
import requests

resp = requests.post(
    "https://jsonplaceholder.typicode.com/posts",
    json={
        "title": "我的第一条帖子",
        "body": "你好，这是我用 Python 发的！",
        "userId": 1
    }
)

print(resp.status_code)   # 201 = 创建成功
data = resp.json()
print(data)
print(f"帖子ID：{data['id']}")
```
