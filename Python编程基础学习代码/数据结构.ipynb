```python
#------列表------
names = ["张三", "李四", "王五"]
print(names[0])      # 张三（第0号，第一个）
print(names[2])      # 王五（第2号，第三个）
print(names[-1])     # 王五（-1 = 倒数第一个）
print(names[1:3])    # ['李四', '王五']  切片：含头不含尾！（和range一样！）
names.append("赵六")        # 增：末尾加一个
names.insert(1, "钱七")     # 增：插到1号位（第二个）
names[0] = "孙八"           # 改：把0号位换成"孙八"
names.remove("李四")        # 删：按"值"删掉李四
last = names.pop()          # 删：按"位置"删最后一个，并把删掉的取出来
print(names)
print(last)
len(names)                  # 查：名单多长
"张三" in names             # 查：张三在不在名单里 → True/False
for name in names:
    print(f"通知：{name}")
    
student = [["小李",11],["小王",22]]
print(student[0])
```

```python
#------字典------
contact = {"name": "张三", "phone": "13800000000", "age": 28}

print(contact["name"])       # 张三（用标签取）
print(contact["phone"])
print(contact["age"])
print(contact.get("age"))    # 28

#print(contact["email"])      # ❌ 报错：没有这个标签
print(contact.get("email"))  # None（get更安全：找不到返回None，不报错）
contact["city"] = "深圳"     # 增：写一个新标签
contact["city"] = "北京" 
contact["time"] = "16:00"
contact["age"] = 29          # 改：标签已存在 → 覆盖
del contact["time"]          # 删
print(contact)

for i,n in contact.items():
    print(f"{i}:{n}")
    
```

```python
#------集合------
s = {1, 2, 2, 3, 3, 3}
print(s)         # {1, 2, 3}  ← 重复的全被自动删掉！
#s[0]            # ❌ 报错：集合不支持索引（它不认第0个是谁）
nums = [6,1, 2, 2, 3, 3, 3,2,3,4,5,6]
unique = list(set(nums))    # [1, 2, 3]
# 列表 → 转集合（自动去重）→ 再转回列表
s.add(11)        # 加一个
s.remove(2)     # 删一个

print(unique)
print(s)

#集合运算
出席A = {"张三", "李四", "王五"}
出席B = {"李四", "王五", "赵六"}

print(出席A & 出席B)    # 交集：两次都出席的 → 李四、王五
print(出席A | 出席B)    # 并集：所有来过的人（自动去重）
print(出席A - 出席B)    # 差集：只在A不在B → 张三
print(出席B - 出席A)    # 差集：只在B不在A → 赵六
```

```python
#------通讯录代码------

phonebook = {} #新建一个空字典存通讯录：{名字：电话}

def add_contact(name,phone):
    phonebook[name] = phone
    print(f"已添加：{name}->{phone}")

def search_contact(name):
    if name in phonebook:
        print(f"查到：{name}->{phonebook[name]}")
    else:
        print(f"通讯录里没有{name}")

def update_phone(name,new_phone):
    if name in phonebook:
        phonebook[name]=new_phone
        print(f"已更新：{name}的电话")
    else:
        print(f"未查到{name},无法更新")

def delete_contact(name):
    if name in phonebook:
        del phonebook[name]
        print(f"已删除{name}的电话")
    else:
        print(f"未查到{name},无法删除")

def show_all():
    print("---通讯录全部---")
    for name,phone in phonebook.items():
        print(f"{name}:{phone}")
        

# ===== 操作一遍 =====
add_contact("张三", "13800000000")
add_contact("李四", "13900000000")
show_all()
search_contact("张三")
update_phone("张三", "13700000000")
#delete_contact("李四")
show_all()
search_contact("王五")

print("---以下是删除全部通讯录---")
def delete_all():
    for name in list(phonebook.keys()):
        delete_contact (name)
delete_all()
#phonebook.clear()   #一键清空
show_all()
```
