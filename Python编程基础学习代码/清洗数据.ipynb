```python
import pandas as pd

df = pd.read_csv("设备故障报表.csv",encoding="gbk")
print(df.head())
print(df.info())
print(df.describe())

print("=======清洗后的结果=======")
#1、看有多少空值
print(df.isnull().sum())

# 2. 删掉空行
df = df.dropna()

# 3. 删掉重复行
df = df.drop_duplicates()

# 4. 改列名
df = df.rename(columns={"修复时长（分钟）":"修复时长"})

# 5. 日期转类型
df["故障时间"] = pd.to_datetime(df["故障时间"],errors = "coerce")

# 6. 再看清洗后的数据

print(df.info())
print(df.shape)


print("======按照筛选条件筛选======")
# 1. 筛选严重故障
print(df[df["故障等级"] == "严重"])

# 2. 筛选修复时长>120分钟的
print(df[df["修复时长(分钟)"] > 120])

# 3. 按修复时长从大到小排
print(df.sort_values("修复时长(分钟)", ascending=False).head(10))


#------分组groupby------
print("======分组groupby======")
# 1. 每个设备故障几次？
print(df.groupby("设备名称")["设备编号"].count())

# 2. 每种故障类型平均修多久？
print(df.groupby("故障类型")["修复时长(分钟)"].mean())

# 3. 每个产线故障几次？
print(df.groupby("产线")["设备编号"].count())



#------透视表------
# 行=设备名称，列=故障等级，值=故障次数
print(pd.pivot_table(df, 
                     index="设备名称", 
                     columns="故障等级", 
                     values="设备编号", 
                     aggfunc="count"))


#------故障排名前十------
top10 = df.groupby("设备名称")["设备编号"].count().sort_values(ascending=False).head(10)
print("=== 设备故障 Top10 ===")
print(top10)

top10 = df.groupby ("故障类型")["处理人"].count ().sort_values (ascending=False).head (10)
print ("=== 故障类型 Top10 ===")
print (top10)       #这里选取“设备编号”也行，选取“处理人”也行，因为只起到一个计数的作用，只要取的列没有空值就对
```

```python
# 删整行（任何一列有空就删）
df = df.dropna()

# 只删"故障时间"为空的行
df = df.dropna(subset=["故障时间"])

# 或者不删，填个默认值
df["产线"] = df["产线"].fillna("未知")
```
