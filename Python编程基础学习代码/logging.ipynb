```python
import logging
import requests

# 配置日志
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filename="weather.log",
    filemode="a"
)

def main():
    logging.info("=== 开始运行 ===")
    
    # 1. 调 API
    try:
        logging.info("正在调用天气 API...")
        resp = requests.get("https://api.open-meteo.com/v1/forecast?latitude=22.5&longitude=114&current=temperature_2m", timeout=10)
        logging.info(f"API 返回状态码: {resp.status_code}")
    except Exception as e:
        logging.error(f"API 调用失败: {e}")
        return
    
    # 2. 处理数据
    try:
        data = resp.json()
        temp = data["current"]["temperature_2m"]
        logging.info(f"当前温度: {temp}°C")
    except Exception as e:
        logging.error(f"数据处理失败: {e}")
        return
    
    # 3. 写文件
    with open("weather.txt", "w", encoding="utf-8") as f:
        f.write(f"深圳当前温度: {temp}°C")
    logging.info("文件已写入 weather.txt")
    
    logging.info("=== 运行完成 ===")

if __name__ == "__main__":
    main()
```
