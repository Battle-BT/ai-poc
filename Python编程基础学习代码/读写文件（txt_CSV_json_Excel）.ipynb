```python
#------写读CSV文件------
#===写===

import csv

with open("data.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["姓名", "年龄", "城市"])
    writer.writerow(["张三", 28, "深圳"])
    writer.writerow(["李四", 32, "北京"])

#===读===

with open("data.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)   # 每行是一个列表
```

```python
#------实际案例（清洗数据）------
import csv
import json

# 先写一个示例 CSV
with open("示例数据.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["姓名", "年龄", "城市", "职业"])
    writer.writerow(["张三", 28, "深圳", "工程师"])
    writer.writerow(["李四", 32, "北京", "产品经理"])
    writer.writerow(["王五", 25, "上海", "设计师"])
    writer.writerow(["赵六", 30, "深圳", "运营"])




# 1. 读 CSV
rows = []
with open("示例数据.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    headers = next(reader)        # 第一行是表头
    for row in reader:
        if not row:               # 跳过空行
            continue
        rows.append(row)

# 2. 清洗：把年龄转成数字
data = []
for row in rows:
    item = {
        "姓名": row[0],
        "年龄": int(row[1]),       # 字符串转数字
        "城市": row[2],
        "职业": row[3]
    }
    data.append(item)

# 3. 写回 JSON
with open("output.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(f"清洗完成，共 {len(data)} 条数据")
for item in data:
    print(item)
```
